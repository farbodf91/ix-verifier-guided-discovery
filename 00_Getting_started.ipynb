# Getting Started with the IX Tutorial
**Open a notebook, connect a Python runtime, and run the cells.**

A notebook combines explanations, executable code cells, and their outputs. Each Colab session runs on its own temporary computer.

## Open the tutorial
[Main teaching notebook](https://colab.research.google.com/github/farbodf91/ix-verifier-guided-discovery/blob/main/01_IX_symbolic_discovery_tutorial.ipynb) · [Cylinder-flow companion](https://colab.research.google.com/github/farbodf91/ix-verifier-guided-discovery/blob/main/02_Cylinder_flow_exploration.ipynb)

1. Choose **Runtime → Change runtime type → Runtime Version → 2026.07** for Python 3.12. CPU is sufficient for the main walkthrough.
2. Run the first setup cell. It retrieves the repository and installs the pinned packages.
3. Run subsequent cells in order using the play button or **Shift+Enter**.

Use **File → Save a copy in Drive** to keep your own edits. Runtime files disappear when the session is deleted; your saved notebook and its outputs remain in Drive.

## What does setup install?
- **NumPy, SciPy, pandas:** numerical arrays, integration, optimization, and tables.
- **Matplotlib:** scientific figures.
- **SymPy:** symbolic expressions.
- **ODEFormer and PyTorch:** the symbolic transformer and its integration interface.

`requirements.txt` specifies compatible versions. The main walkthrough uses saved candidate equations and computes their checks and predictions live. Optional generation loads the pretrained checkpoint automatically when first requested.

## If setup is interrupted
| What you see | What to do |
|---|---|
| Python version message | Select runtime 2026.07 and reconnect. |
| Packages changed after imports | Restart the session and run from setup again. |
| Runtime disconnected | Reconnect, rerun setup, then run the earlier cells in order. |
| Missing variable | Run preceding cells; the kernel stores variables only during the current session. |
| Old code after a repository update | Start a fresh Colab session and reopen the updated notebook. |

Figures and formatted equation systems are saved as notebook outputs. Rerunning their cells refreshes them.

## Local setup (optional)
Use Python 3.12 in a separate environment. Clone the [repository](https://github.com/farbodf91/ix-verifier-guided-discovery), then open a terminal in its folder:

```bash
python3.12 -m venv .venv
source .venv/bin/activate
python -m pip install -r requirements.txt jupyterlab
jupyter lab
```

On Windows, activate with `.venv\Scripts\activate`. In VS Code, open the repository folder and select the environment's Python kernel. Run the notebook's setup cell before continuing.

The `tutorial_lib` folder contains ordinary Python files. Open them directly or inspect a function with `inspect.getsource`; editor navigation shortcuts depend on the editor configuration.
